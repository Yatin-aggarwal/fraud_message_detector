```python
label_map = {'ham': 0, 'smish': 1}
df['label'] = df['label'].map(label_map)
```

```python
label_map = {'normal': 0, 'scam': 1}
df2['label'] = df2['label'].map(label_map)
```

```python
df3 = pd.concat([df,df2], axis = 0 )
```

```python
len(df3), len(df), len(df2)
```

```python
device = torch.device("cuda" if torch.cuda.is_available else "cpu")
```

```python
model_name = "distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name )
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=1).to(device)
```

```python
model
```

```python
epochs = 10
batch_size = 32 
```

```python
data = tokenizer(list(df3['message']), padding=True, truncation=True, max_length=128, return_tensors="pt")
```

```python
data
```

```python
labels = torch.tensor(df3['label'].values, dtype=torch.float)  # float for BCE, long for CE

dataset = TensorDataset(data['input_ids'], data['attention_mask'], labels)

val_ratio = 0.2  #
val_size  = int(val_ratio * len(dataset))
train_size = len(dataset) - val_size


generator = torch.Generator().manual_seed(42)
train_ds, val_ds = random_split(dataset, [train_size, val_size], generator=generator)
train = DataLoader(train_ds, batch_size=batch_size, shuffle=True)
val  = DataLoader(val_ds, batch_size=batch_size, shuffle=False)
```

```python
optimizer = AdamW(model.parameters(), lr=2e-5, weight_decay=0.01)

# Scheduler
num_training_steps = len(train) * epochs
num_warmup_steps = int(0.1 * num_training_steps)

scheduler = get_linear_schedule_with_warmup(optimizer, num_warmup_steps, num_training_steps)

loss_fn = BCEWithLogitsLoss()
```

```python
model.train()

for epoch in range(epochs):
    running_loss = 0.0

    for input_ids, attention_mask, label in train:
        
        input_ids = input_ids.to(device)
        attention_mask = attention_mask.to(device)
        label = label.to(device)

        optimizer.zero_grad()

        outputs = model(input_ids=input_ids, attention_mask=attention_mask)
        logits = outputs.logits  
        loss = loss_fn(logits, label.float().unsqueeze(1))
        loss.backward()

        
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

        optimizer.step()
        scheduler.step()  

        running_loss += loss.item()
    avg_loss = running_loss / len(train)
    print(f"Epoch {epoch+1}/{epochs} - train loss: {avg_loss:.4f}")
```

```python
Y_pred = np.asarray([])
Y_acc = np.asarray([])
with torch.no_grad():
    for input_ids, attention_mask, label in val:
        input_ids = input_ids.to(device)
        attention_mask = attention_mask.to(device)
        label = label.to(device)
        outputs = model(input_ids=input_ids, attention_mask=attention_mask)
        logits = outputs.logits
        y_pred = torch.sigmoid(logits).cpu().squeeze(1).numpy()>=0.5
        y_pred = np.where(y_pred == True, 1, y_pred)
        y_pred = np.where(y_pred == False, 0, y_pred)
        y = label.cpu().numpy()
        Y_pred = np.append(Y_pred, y_pred)
        Y_acc= np.append(Y_acc, y)
```

```python
cm = confusion_matrix(Y_pred, Y_acc)
```
